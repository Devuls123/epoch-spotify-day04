# 10. Now create at least 6 visualizations

## Visualization 1 — Distribution of Popularity

```python
plt.figure(figsize=(10, 6))

sns.histplot(data=df, x='popularity', bins=20, kde=True)

plt.title('Distribution of Spotify Track Popularity')
plt.xlabel('Popularity Score')
plt.ylabel('Number of Tracks')

plt.show()
```

Insight

The popularity distribution shows how Spotify tracks are distributed across popularity scores. The visualization helps identify whether most tracks have low, medium, or high popularity and whether the distribution is concentrated around a particular range.

# Visualization 2 — Top 10 Music Genres

```python
top_genres = df['track_genre'].value_counts().head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index
)

plt.title('Top 10 Music Genres by Number of Tracks')
plt.xlabel('Number of Tracks')
plt.ylabel('Genre')

plt.show()
```

Insight

The bar chart shows the genres with the highest number of tracks in the dataset. It highlights which music categories are most represented and provides an overview of the dataset's genre composition.

#Visualization 3 — Energy vs Popularity

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='energy',
    y='popularity',
    alpha=0.4
)

plt.title('Relationship Between Energy and Popularity')
plt.xlabel('Energy')
plt.ylabel('Popularity')

plt.show()
```

Insight

The scatter plot helps examine whether energetic songs tend to have higher popularity. If the points are widely scattered, it suggests that energy alone may not strongly determine a track's popularity

# Visualization 4 — Audio Features Correlation

```python
correlation = df[audio_features].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f'
)

plt.title('Correlation Between Spotify Audio Features')

plt.show()
```

Insight

The correlation heatmap reveals relationships between different audio features. Strong positive or negative correlations indicate features that tend to change together, while values close to zero suggest weak relationships.

# Visualization 5 — Popularity by Top Genres

```python
top_10_genres = df['track_genre'].value_counts().head(10).index

genre_data = df[df['track_genre'].isin(top_10_genres)]
```

```python
plt.figure(figsize=(14, 7))

sns.boxplot(
    data=genre_data,
    x='track_genre',
    y='popularity'
)

plt.title('Popularity Distribution Across Top 10 Genres')
plt.xlabel('Genre')
plt.ylabel('Popularity')

plt.xticks(rotation=45)

plt.show()
```

Insight

The box plot compares the popularity distribution across the most represented genres. It shows differences in median popularity, variation, and potential outliers between genres.

# Visualization 6 — Danceability vs Energy

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='danceability',
    y='energy',
    alpha=0.4
)

plt.title('Relationship Between Danceability and Energy')
plt.xlabel('Danceability')
plt.ylabel('Energy')

plt.show()
```
